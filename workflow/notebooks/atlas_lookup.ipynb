```python
import pandas
import pathlib

from alphagenome.atlas import atlas
from alphagenome.data import genome

ROOT = pathlib.Path('/storage/mathelierarea/processed/eirikhoy/action_var')

API_KEY = (ROOT / 'resources/API_KEY').read_text().strip()

ATLAS_SCORERS = ['AVI_SCORE', 'RNA_SEQ', 'SPLICE_SITES', 'SPLICE_SITE_USAGE', 'SPLICE_JUNCTIONS',
                 'POLYADENYLATION', 'CAGE', 'PROCAP', 'DNASE', 'ATAC', 'CHIP_HISTONE', 'CHIP_TF',
                 'CONTACT_MAPS']

client = atlas.create(API_KEY)
```

```python
# 1. See which scorers exist
scorers = list(client.scorer_metadata().keys())
print(scorers)
```

```python
# 2. Query every SNV in a small interval
interval = genome.Interval(chromosome="chr22", start=36_201_000, end=36_201_100)

results = client.query_interval(
    interval, 
    requested_scorers=[scorers[0]],
)
```

```python
adata = results[scorers[0]]
print(adata)
print(adata.obs.head())
print(adata.var.head())
```
